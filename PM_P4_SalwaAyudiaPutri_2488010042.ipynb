```python
import pandas as pd
import numpy as np

data = {
    'usia': [19, 22, np.nan, 45, 17, 42, 20, np.nan, 18, 30],
    'pendapatan': [5_000_000, 8_300_000, 7_500_000, 20_000_000, 5_700_000, 19_000_000, 9_100_000, 3_300_000, np.nan, 10_900_000],
    'pendidikan': ['SMA', 'S1', 'SMA', 'S2', 'SMA', 'S2', 'S1', 'S1', 'SMA', 'S2'],
    'kota': ['Bandung', 'Cirebon', 'Bandung', 'Indramayu', 'Cirebon', 'Indramayu', 'Cirebon', 'Bandung', 'Bandung', 'Cirebon'],
    'membeli': ['Tidak', 'Ya', 'Tidak', 'Ya', 'Tidak', 'Ya', 'Ya', 'Tidak', 'Tidak', 'Ya'],
}

df = pd.DataFrame(data)
df
```

```python
df_median = df.copy()
```

# Latihan Mandiri

1. Ganti imputasi usia menjadi mean; bandingkan hasilnya.

```python
df_median['usia'] = df_median['usia'].fillna(df_median['usia'].median())
df_median['pendapatan'] = df_median['pendapatan'].fillna(df_median['pendapatan'].median())

print("Mean usia:", df_median['usia'].median())
print("Mean pendapatan:", df_median['pendapatan'].median())
print(display(df_median))
```

```python
df['usia'] = df['usia'].fillna(df['usia'].mean())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].mean())

print("Mean usia:", df['usia'].mean())
print("Mean pendapatan:", df['pendapatan'].mean())
print(display(df))
```

#### Perbandingan Hasil:
Pada imputasi median semua nilai kosong (NaN) baik di kolom usia maupun pendapatan digantikan dengan nilai tengah, sedangkan pada imputasi mean semua nilai kosong nya digantikan oleh nilai rata-rata. Sehingga nilai pengganti untuk NaN berbeda antara median dan mean.


3. Tambahkan kolom kategorikal baru dan terapkan one-hot encoding.

```python
X = df.drop(columns=['membeli'])
y = df['membeli']

X['pendidikan'] = X['pendidikan'].map({'SMA':0,'S1':1,'S2':2})
X = pd.get_dummies(X, columns=['kota'], dtype=int)
y = y.map({'Tidak':0,'Ya':1})

print("Hasil encoding X:")
display(X)

print("\n\nHasil encoding y:")
display(y)
```

2. Gunakan MinMaxScaler (0–1) sebagai ganti StandardScaler; tampilkan describe().

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
X, y, test_size=0.3, random_state=42)


from sklearn.preprocessing import MinMaxScaler
num = ['usia','pendapatan','pendidikan']
sc = MinMaxScaler()

X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num])

print("Hasil X_train setelah MinMaxScaler:")
display(X_train)

print("\n\nHasil X_test setelah MinMaxScaler:")
display(X_test)

print("\n\nDescribe X_train:")
display(X_train[num].describe())
```

4. Jelaskan mengapa fit_transform pada latih tetapi hanya transform pada uji.


*  Jawab: <br>
Karena fit_transform digunakan untuk mempelajari kategori yang unik dan mengubah nya menjadi kolom biner (0 dan 1) agar komputer dapat memprosesnya, sedangkan data uji hanya menggunakan transform saja. Ini bertujuan untuk menjaga konsistensi jumlah dan urutan kolom agar model nya tidak error dan mencegah terjadinya kebocoran data (data leakage).



# Refleksi

1. Mengapa urutan 'split dulu, baru scaling' penting?


*   Jawab: <br>
Karena jika penskalaan (scaling) di fit pada seluruh data sebelum di split, informasi data uji nya akan bocor ke pelatihan. Akibatnya evaluasi menjadi tidak jujur dan model tampak lebih baik daripada kenyataanya.



2. Kapan label encoding dan kapan one-hot encoding?


*   Jawab: <br>
Label encoding cocok digunakan saat data nya adalah data ordinal (berurutan), karena mengubah teks menjadi angka berurutan yang hasilnya adalah satu kolom angka. <br>
One-hot encoding cocok digunakan untuk data nominal (tanpa urutan) dengan cara membuat kolom biner baru (0 dan 1) sehingga kolom akan membengkak.

3. Apa perbedaan normalisasi dan standardisasi?


*   Jawab: <br>
Normalisasi (Min-Max) berguna bila butuh batas yang jelas, skala ke rentang tetap 0-1, dan sensitif terhadap outlier. Sedangkan Standardisasi (Z-core) tidak membatasi rentang, lebih tahan outlier serta mengubah mean (rata-rata) = 0 dan standar deviasi = 1.
